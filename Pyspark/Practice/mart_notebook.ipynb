## Schema

```python
df.printSchema()
```

```python
my_ddl_schema='''
 Item_Identifier STRING,
 Item_WeighT string,
 Item_Fat_Content string,
 Item_Visibility double,
 Item_Type string,
 Item_MRP double,
 Outlet_Identifier string,
 Outlet_Establishment_Year integer,
 Outlet_Size string,
 Outlet_Location_Type string,
 Outlet_Type string,
 Item_Outlet_Sales double
'''
```

```python
df=spark.read.format('csv')\
    .schema(my_ddl_schema)\
        .option('header',True)\
            .load('BigMart Sales.csv')
```

```python
df.printSchema()
```

### StructType() Schema

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
my_struct_schema=StructType([
    StructField('Item_Identifier',StringType(),True),
    StructField('Item_WeighT string',StringType(),True),
 StructField('Item_Fat_Content string',StringType(),True),
 StructField('Item_Visibility double',StringType(),True),
 StructField('Item_Type string',StringType(),True),
 StructField('Item_MRP double',StringType(),True),
 StructField('Outlet_Identifier string',StringType(),True),
 StructField('Outlet_Establishment_Year integer',StringType(),True),
 StructField('Outlet_Size string',StringType(),True),
 StructField('Outlet_Location_Type string',StringType(),True),
 StructField('Outlet_Type string',StringType(),True),
 StructField('Item_Outlet_Sales double',StringType(),True)
])
```

```python
df=spark.read.format('csv')\
    .schema(my_struct_schema)\
        .option('header',True)\
            .load('BigMart Sales.csv')
```

```python
df.printSchema()
```

## Date Functions

Current_Date

```python
df = df.withColumn('curr_date',current_date())

df.show()
```

Date_Add()

```python
df = df.withColumn('week_after',date_add('curr_date',7))

df.show()
```

Date_Sub()

```python
df.withColumn('week_before',date_sub('curr_date',7)).show()
```

```python
df = df.withColumn('week_before',date_add('curr_date',-7)) 

df.show()
```

DateDIFF

```python
df = df.withColumn('datediff',datediff('week_after','curr_date'))

df.show()
```

Date_Format()

```python
df = df.withColumn('week_before',date_format('week_before','dd-MM-yyyy'))

df.show()
```
